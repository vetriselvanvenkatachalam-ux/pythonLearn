DataType


1) Numeric Type

```python
"""INTEGER"""
A = 5 ;
print(A);
print(type(A))

"""FLOAT"""
B=23.3
print(B)
print(type(B));

"""COMPLEX"""
C=4+4J
print(C)
print(type(C))
# Imaginary Part 
print(C.imag)
# Real Part
print(C.real);
print(type(C))
```

2.Sequence Type

```python
"""TUPLE"""
A = (1,2,3,4,5);
print(A);
print(type(A));

"""SET"""
B = {1,2,222,3,3,4,4,5,6,6}
print(B)
print(type(B))

"""dict"""
C={'key1' : 'Value',
   'Key2' : '1',
   'key3' : None ,
   'key4' : 'True'
   };
print(C)
print(type(C))
```

3. Bollean


```python
"""BOOLEAN"""
A = True
B = False
print(A)
print(B)
print(type(A))
print(type(B))
```

4 . Special Type

```python
"""NONE TYPE"""
A = None
print(A)
print(type(A))
```

5 . Escape Character

```python
"""New Line"""
A = "Hello\nWorld"
B = "Hello\rWorld"
C = "Hello\tWorld"
D = "Hello\"World\""
E = '\'Hello\'World'

print(A)
print(B)
print(C)
print(D)
print(E)
```

6. Float Power 10 

```python
"""Float Power"""

A = 10e3
print(A)
print(type(A))
```
